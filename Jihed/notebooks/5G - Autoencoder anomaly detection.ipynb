# 5.3 Best-Effort Anomaly Detection - Autoencoder (MLP reconstruction)Semi-supervised approach: learn normal best-effort behavior, then detect premium-in-best-effort anomalies by reconstruction error.

```python
from pathlib import Pathimport numpy as npimport pandas as pdfrom sklearn.model_selection import train_test_splitfrom sklearn.neural_network import MLPRegressorfrom sklearn.preprocessing import StandardScalerfrom sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matriximport matplotlib.pyplot as pltimport seaborn as sns
```

```python
cwd = Path.cwd().resolve()if cwd.name == 'notebooks' and (cwd.parent / 'data').exists():    project_dir = cwd.parentelif (cwd / 'data').exists() and (cwd / 'notebooks').exists():    project_dir = cwdelif (cwd / 'Jihed' / 'data').exists():    project_dir = cwd / 'Jihed'else:    raise FileNotFoundError(f'Unable to locate project directories from {cwd}')ROOT = project_dirTRAIN_PATH = ROOT / 'data' / 'processed' / 'train_anomaly_detection.csv'train = pd.read_csv(TRAIN_PATH)print('train shape:', train.shape)
```

```python
feature_cols = [    'packet_delay_ms',    'packet_loss_rate',    'packet_loss_log10',    'time_sin',    'time_cos',    'critical_service_count',    'is_gbr',    'is_non_gbr',    'is_5g',    'lte_5g_category',]X = train[feature_cols].copy()y_true = train['anomaly_premium_in_best_effort'].astype(int)baseline_mask = (train['is_best_effort_slice'] == 1) & (train['premium_intent'] == 0)idx_train, idx_val = train_test_split(    train.index,    test_size=0.30,    random_state=42,    stratify=y_true,)base_train_mask = baseline_mask.loc[idx_train]X_fit = X.loc[idx_train][base_train_mask]X_val = X.loc[idx_val]y_val = y_true.loc[idx_val].to_numpy()print('baseline rows used for training:', len(X_fit))
```

```python
scaler = StandardScaler()X_fit_scaled = scaler.fit_transform(X_fit)X_all_scaled = scaler.transform(X)X_val_scaled = scaler.transform(X_val)candidates = [    (8, 4, 8),    (12, 6, 12),    (16, 8, 16),    (24, 12, 24),    (32, 16, 32),]alphas = [1e-5, 1e-4, 1e-3]best = Nonefor hidden in candidates:    for alpha in alphas:        model = MLPRegressor(            hidden_layer_sizes=hidden,            activation='relu',            solver='adam',            alpha=alpha,            max_iter=1200,            random_state=42,        )        model.fit(X_fit_scaled, X_fit_scaled)        val_recon = model.predict(X_val_scaled)        val_err = np.mean((X_val_scaled - val_recon) ** 2, axis=1)        local_best = None        for q in np.linspace(0.50, 0.99, 250):            thr = float(np.quantile(val_err, q))            pred = (val_err >= thr).astype(int)            f1 = f1_score(y_val, pred)            if local_best is None or f1 > local_best[0]:                local_best = (f1, q)        f1, q = local_best        if best is None or f1 > best[0]:            best = (f1, hidden, alpha, q, model)best_f1_val, best_hidden, best_alpha, best_q, autoencoder = bestprint('best config:', {'hidden': best_hidden, 'alpha': best_alpha, 'q': round(float(best_q), 4), 'f1_val': round(float(best_f1_val), 4)})
```

```python
recon_all = autoencoder.predict(X_all_scaled)train['ae_reconstruction_error'] = np.mean((X_all_scaled - recon_all) ** 2, axis=1)threshold = float(np.quantile(train['ae_reconstruction_error'], best_q))train['ae_threshold'] = thresholdtrain['ae_pred'] = (train['ae_reconstruction_error'] >= threshold).astype(int)metrics = pd.DataFrame([    {        'accuracy': accuracy_score(y_true, train['ae_pred']),        'precision': precision_score(y_true, train['ae_pred'], zero_division=0),        'recall': recall_score(y_true, train['ae_pred'], zero_division=0),        'f1_score': f1_score(y_true, train['ae_pred'], zero_division=0),    }])display(metrics.round(4))cm = confusion_matrix(y_true, train['ae_pred'])fig, ax = plt.subplots(figsize=(6, 4.8))sns.heatmap(cm, annot=True, fmt='d', cmap='Greens', cbar=False, linewidths=0.5, linecolor='white', ax=ax)ax.set_title('Confusion Matrix - Autoencoder MLP')ax.set_xlabel('Predicted')ax.set_ylabel('Actual')ax.set_xticklabels(['Normal', 'Anomaly'])ax.set_yticklabels(['Normal', 'Anomaly'], rotation=0)plt.tight_layout()plt.show()
```
